## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Search Question

Which webpages should be prioritized for content review and refresh?

### Decision

Determine which pages should receive limited SEO and content optimization resources first.

### Action

SEO and content specialists review the highest-ranked pages and decide whether to refresh, expand, rewrite, monitor, or leave them unchanged.

### Unit of Analysis

One webpage (one content item).

### Output

A ranked list of webpages with an opportunity score and supporting reason codes.

### Cost of a Wrong Recommendation

A false positive wastes time by recommending pages that are unlikely to improve.

A false negative ignores pages with real improvement potential, resulting in missed traffic and business opportunities.

### Why can data or ML help?

Search performance depends on several interacting factors such as visibility, ranking position, freshness, and engagement. Machine learning can help identify useful patterns across these signals and prioritize pages more effectively than simple manual rules.

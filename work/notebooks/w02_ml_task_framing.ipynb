## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

I chose **Lane 2: Refresh / Content Opportunity Scoring**. This project is primarily a **ranking and scoring** task.

Instead of making a simple yes/no decision, the goal is to assign an opportunity score to each webpage and rank pages from highest to lowest priority. Pages with higher scores should be reviewed first because they are more likely to benefit from content updates.

This framing matches the real business problem because SEO teams usually have limited resources and need a prioritized review queue rather than a binary prediction.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

For the starter dataset, I will use the provided proxy label:

**is_declining_label = (trend_direction == "down")**

This label indicates whether a page is currently showing signs of declining performance. Although it is not a future outcome, it provides a useful starting point for developing and evaluating ranking methods.

For the final capstone, I plan to use a stronger future-looking label, such as predicting whether a page will decline or recover during a future time window while avoiding data leakage.

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

df["is_declining_label"] = (df["trend_direction"] == "down").astype(int)

df[["content_id","trend_direction","is_declining_label"]].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary evaluation metric will be **Precision@K**, especially **Precision@50**.

This metric measures how many of the top-ranked pages are actually useful review candidates. It aligns well with the real business workflow because SEO specialists usually review only a limited number of pages at a time.

I will also consider Average Precision and Recall when comparing different models.

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

The unit of analysis is **one webpage (one content item)**.

Each row in the dataset represents one webpage with search, engagement, freshness, and content features. The model will produce one opportunity score for each page, allowing pages to be ranked according to their expected review priority.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Simple rules such as selecting only stale pages or only pages with high impressions cannot capture the complex interactions between search visibility, ranking position, engagement, freshness, and demand.

Machine learning can learn patterns across multiple features simultaneously and generate more accurate rankings than fixed rule-based systems while still supporting human decision-making.
